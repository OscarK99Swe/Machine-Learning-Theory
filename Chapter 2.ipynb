## Q8, explain what the code down below does. Why is it important to be able to save a model?

```python
#creates a synthetic regression dataset
from sklearn.datasets import make_regression 
#Initializes a Linear Regression model and trains it to learn the 
#relationship between features (X) and targets (Y)
from sklearn.linear_model import LinearRegression 
#Saves the trained model by seralizing it and saves it locally 
from joblib import dump, load

X, y = make_regression(n_samples=20000, n_features=3,
noise=0.1)
model = LinearRegression().fit(X, y)
dump(model, "linear_model.joblib")
model_loaded = load("linear_model.joblib")
#Uses the newly loaded model to calculate and display the first 5 rows of X
print(model_loaded.predict(X[:5]))
```

***Why is it important to save model?***

Lots of reasons, but the main one is to not have to re-train a model 
each time you want to use it again. Having to retrain a model might give different
results on each new iteration, thus limiting stable returns.

## Q9 

```python
#A 

import pandas as pd

# Load the dataset from the Dataset folder
df = pd.read_csv("Datasets/data_01.csv")

# Display the first few rows to verify it loaded correctly
df.head()
```

```python
#B

X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

```python
#C 

from sklearn.model_selection import train_test_split

# 1. First split: 20% test data, 80% remaining for train/val
X_train_full, X_test, y_train_full, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# 2. Second split: 15% of the REMAINING data for validation, 85% for training
X_train, X_val, y_train, y_val = train_test_split(
    X_train_full, y_train_full, test_size=0.15, random_state=42
)

# Optional: Print shapes to verify proportions
print(f"Total samples: {len(X)}")
print(f"Train samples: {len(X_train)}")
print(f"Validation samples: {len(X_val)}")
print(f"Test samples: {len(X_test)}")
```

```python
#D

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor

# Initialize the models
lr_model = LinearRegression()
dt_model = DecisionTreeRegressor(random_state=42)

# Train both models on the training data
lr_model.fit(X_train, y_train)
dt_model.fit(X_train, y_train)
```

```python
#E 

from sklearn.metrics import mean_squared_error

# Generate predictions on the validation set
lr_val_preds = lr_model.predict(X_val)
dt_val_preds = dt_model.predict(X_val)

# Evaluate using Mean Squared Error (MSE) - lower is better
lr_mse = mean_squared_error(y_val, lr_val_preds)
dt_mse = mean_squared_error(y_val, dt_val_preds)

print(f"Linear Regression Validation MSE: {lr_mse:.4f}")
print(f"Decision Tree Validation MSE: {dt_mse:.4f}")
```

```python
#F

# Programmatically select the best model based on validation MSE
if lr_mse < dt_mse:
    print("Linear Regression performed better. Retraining...")
    best_model = LinearRegression()
else:
    print("Decision Tree performed better. Retraining...")
    best_model = DecisionTreeRegressor(random_state=42)

# Retrain the chosen model on the combined train and validation set (X_train_full, y_train_full)
best_model.fit(X_train_full, y_train_full)
```

```python
#G

# Evaluate the newly trained best_model on the unseen test data
test_preds = best_model.predict(X_test)
test_mse = mean_squared_error(y_test, test_preds)

print(f"Final Model Test MSE: {test_mse:.4f}")
```

```python
#H

# Create a final production model and train it on the entire original dataset (X, y)
final_production_model = type(best_model)(**best_model.get_params())
final_production_model.fit(X, y)

print("Model successfully retrained on the entire dataset.")
```

## Q 10

```python
#A 

import pandas as pd

# Load the dataset
df = pd.read_csv("Datasets/salary_dataset.csv")

# Dependent variable (y) is 'Salary' (what we want to predict)
# Independent variable (X) is 'YearsExperience'
X = df[["YearsExperience"]]  # Kept as a 2D DataFrame for Scikit-Learn
y = df["Salary"]

# Display the first few rows
df.head()
```

```python
#B

from sklearn.model_selection import train_test_split

# Split into training (80%) and test (20%) sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print(f"Training samples: {len(X_train)}")
print(f"Test samples: {len(X_test)}")
```

```python
#C

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_validate
from sklearn.tree import DecisionTreeRegressor

# Initialize two regression models
lr_model = LinearRegression()
dt_model = DecisionTreeRegressor(random_state=42)

# Perform 5-fold cross-validation using neg_root_mean_squared_error
cv_results_lr = cross_validate(
    lr_model,
    X_train,
    y_train,
    cv=5,
    scoring="neg_root_mean_squared_error",
)
cv_results_dt = cross_validate(
    dt_model,
    X_train,
    y_train,
    cv=5,
    scoring="neg_root_mean_squared_error",
)

# Convert negative RMSE scores to positive mean RMSE
lr_cv_rmse = -np.mean(cv_results_lr["test_score"])
dt_cv_rmse = -np.mean(cv_results_dt["test_score"])

print(f"Linear Regression Cross-Validation RMSE: {lr_cv_rmse:.2f}")
print(f"Decision Tree Regressor Cross-Validation RMSE: {dt_cv_rmse:.2f}")
```

```python
#D

from sklearn.metrics import mean_squared_error

# Select the best performing model based on lower CV RMSE
if lr_cv_rmse < dt_cv_rmse:
    print("Linear Regression performed better in CV. Training on full train set...")
    best_model = lr_model
else:
    print(
        "Decision Tree Regressor performed better in CV. Training on full train set..."
    )
    best_model = dt_model

# Fit the chosen model on the full training set
best_model.fit(X_train, y_train)

# Evaluate on the unseen test set
y_pred = best_model.predict(X_test)
test_rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print(f"Test Set RMSE: {test_rmse:.2f}")
```
